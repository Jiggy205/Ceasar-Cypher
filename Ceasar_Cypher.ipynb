```python
def encode(message, shift):
    encoded = []
    message = message.lower()
    for ch in message:
       if ch.isalpha():
        x = ord(ch) #Ai help to understand how to use. converts letter no num
        x += shift #shifts the number
        while x > 122:  #if it is past z
          x -= 26
        encoded.append(chr(x)) #Ai help to understand how to use
       else:
        encoded.append(ch)
    return "".join(encoded)
def decode(message, shift):
  encoded = []
  for ch in message:
       if ch.isalpha():
        x = ord(ch)
        x -= shift #shifts the number back
        while x < 96: #if it is below a
          x += 26
        encoded.append(chr(x))
       else:
        encoded.append(ch)
  return "".join(encoded) #Ai helped because i wasnt getting the right result

message = input("What is the message? ")
z = 0
while z == 0:
  try:
      shift = int(input("How much do you want to shift? "))
      z +=1
  except ValueError:
      print("please give a number")


option = input("encode or decode? ")
if option == "decode":
  print(decode(message, shift))
elif option == "encode":
  print(encode(message, shift))


```
